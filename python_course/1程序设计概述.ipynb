## 1. 程序设计语言

+ **程序**
    - 程序是一组计算机指令序列，可以执行某些操作或运算，从而解决具体问题。  
+ **程序设计语言**
    - 程序设计语言是用来编写程序的形式化语言，是一套符号和规则系统，定义了语法（结构）和 语义（含义）。   
+ **程序设计语言分类**

    - 由低级到高级，分为机器语言、汇编语言、高级语言，越低级越接近二进制指令，越高级越接近人类思维。
    - 高级语言有：C、C++、fortran、java、matlab、php、R、python 等。
    - 通用和专用编程语言
        + 通用编程语言指用于编写多种用途程序的编程语言，如c、c++、java、python等
        + 专用编程语言指专用特定领域的编程语言，如 html(网页)、javascript、matlab(科学计算)、sql(数据库)
+ **程序执行方式**
    - 有编译执行和解释执行两种方式
    - 编译执行将源程序整体翻译成一个可执行的目标程序(二进制指令)，如c、c++
    - 解释执行将源程序逐句解释成二进制指令，解释一句执行一句，不生成可执行文件，速度慢。如python、basic
    - 采用编译执行的语言也称静态语言，采用解释执行的语言称为动态语言或脚本语言。
+ **python 语言介绍**
    - 发展历史：诞生于1991年，开源开放为设计理念，python 2.x 已成遗产，python 3.x 是现在和未来。
    - 主要特点：语法简洁、开源免费、平台无关、生态丰富、编程模式多样（支持面向过程和面向对象编程）。
    - 应用领域：科学计算、统计分析、人工智能、数据处理与分析、游戏开发、web开发、网络爬虫等
    

## 2. 程序设计方法

+ 程序执行IPO模式：即输入数据(Input)、处理(Process)、输出结果(Output)
+ 程序设计的本质：利用计算机解决实际问题的过程。
+ 程序设计过程：
   
|阶段|过程步骤|概要|任务|
|--|--|:--|:--|
|分析|定义问题|做什么？|深入分析实际问题，准确提炼、描述要解决的问题|
|设计|建立模型|如何做？（理论准备）|建立数学、统计、物理模型，或机器学习模型等|
|设计|设计算法|如何做？（具体步骤）|将模型问题分解成计算机能够执行的若干流程和具体步骤|
|实现|编写程序|动手做，编程|选择合适编程语言实现算法|
|实现|测试运行|动手做，测试|程序测试、运行、升级或维护等|
    

## 3. 算法概念和特性

+ <mark>算法</mark>是解决问题所采取的方法和操作步骤，操作必须是计算机能够执行的，比如加减乘除等可以，吃饭理发就不可以。
+ 算法与程序是两个相关但不同的概念，算法强调计算所采用的方法和步骤，也称为计算流程或计算方法，与编程语言无关；而程序是算法的实现，是计算机指令序列，更具体，与编程语言相关。
+ 一个问题可以有多种算法实现。例如：1 + 2 + 3 + ... + 100, 可以有两种算法实现。
+ 算法的特性：有穷性、确定性、输入或输出、可行性。
+ 算法的表示法
    - 自然语言：优点是简单，缺点是人类语言描述问题容易产生歧义。
    - 伪代码：介于自然语言和程序语言之间的文字和符号来描述算法。
    - 流程图：使用图形符号表示各种操作，用箭头表示语句的执行顺序。
+ 算法的基本控制结构：顺序结构、分支结构和循环结构

## 4. 程序设计工具

+ python 语言可以在多个平台上运行，如 windows、Linux 或 mac os等。
+ python 语言有很多优秀的集成开发环境（Integrated Development Environment,IDE）,辅助完成程序开发工作。
+ 常用的Python IDE 有：
    * IDLE: python 标准发行版自带，简单小巧，包括交互命令行、编辑器和调试器等组件。
    * PyCharm: 商业软件，JetBrains推出的功能全面的面向专业开发的python IDE。
    * Anaconda： 免费软件，“python开发大礼包，全家桶”，主要包含
        - Jupyter notebook，基于网页的交互式计算环境，支持代码、文字和可视化的混合编辑和实时运行，适合数据科学。
        - spyder,专为科学计算设计的IDE，侧重数据科学工作流。
        - Ipython，增强型交互式命令行工具。
    * Visual Studio Code (VScode): 微软开发的轻量级跨平台代码编辑器，通过丰富扩展支持多语言灵活开发。
+ python 程序运行两种方式：交互式和文件式。
    * 交互式：在交互提示符 >>> 后输入命令，按 Enter 键后下一行给出运行结果。非常规运行方式。如上 Ipython
    * 文件式：将多行程序一次性写入文件中，然后批量执行文件中全部代码。最常用执行方式，如上IDLE, PyCharm, spyder等
    * 注意：jupyter notebook 结合交互式和文件式程序运行的优点，编程方式灵活。将作为本教程使用的 IDE 工具。
+ 集成开发环境（IDE）的基本操作（需掌握jupyter notebook 和 spyder IDE）：
    - 下载、安装和配置、启动和关闭等
    - 新建、保存、重命名、拷贝、删除 python 源文件等
    - 编辑、运行、调试 python 源文件

### 注释
+ 注释是程序员在代码中加入的辅助性文字，用来提高代码的可读性。
+ 注释文字会被解释器过滤掉，不被执行。
+ 两种注释方法：单行注释和多行注释。单行注释以#开头，多行注释以'''(3个单引号)开头和结尾。

如下例所示：

````python
# 这是单独注释行
print("Hello world") # 这也是注释，附在语句行后面

'''
多行注释中第一行
多行注释中第二行
'''
print("I like Python")
````

+ 数据输出函数 print()，如 print('hello')中，'hello’为输入参数，print函数在输出设备上显示数据，但函数本身无返回值。
